# 3. PRE-PROCESSING

3.1 The variable "Time" does not provide any information. We will remove it.

```python

from sklearn.preprocessing import StandardScaler
datos.drop(['Time'], axis=1, inplace=True)
datos['Amount'] = StandardScaler().fit_transform(datos['Amount'].values.reshape(-1,1))

from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(datos, test_size=0.2, random_state=42)
X_train = X_train[X_train.Class == 0]
X_train = X_train.drop(['Class'], axis=1)
X_train = X_train.values

Y_test = X_test['Class']
X_test = X_test.drop(['Class'], axis=1)
X_test = X_test.values
```

# 4. AUTOENCODER: 29-20-14-20-29, tanh-relu-tanh-relu

```python
import numpy as np
np.random.seed(5)
from keras.models import Model, load_model
from keras.layers import Input, Dense
from keras.callbacks import EarlyStopping


dim_entrada = X_train.shape[1]          # 29
capa_entrada = Input(shape=(dim_entrada,))

encoder = Dense(20, activation='tanh')(capa_entrada)
encoder = Dense(14, activation='relu')(encoder)

decoder = Dense(20, activation='tanh')(encoder)
decoder = Dense(29, activation='relu')(decoder)

autoencoder = Model(inputs=capa_entrada, outputs=decoder)

from keras.optimizers import SGD
sgd = SGD(learning_rate=0.1)
autoencoder.compile(optimizer='sgd', loss='mse')

nits = 30 # Reduced number of epochs
tam_lote = 32

early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

autoencoder.fit(X_train, X_train, epochs=nits, batch_size=tam_lote, shuffle=True, validation_data=(X_test,X_test), verbose=1, callbacks=[early_stopping])
```

# 5. VALIDACIÓN


Predicción X_test -> Autoencoder -> X_pred

```python
X_pred = autoencoder.predict(X_test)
ecm = np.mean(np.power(X_test-X_pred,2), axis=1)
print(X_pred.shape)

# Precision-recall graph to determine the threshold
from sklearn.metrics import confusion_matrix, precision_recall_curve
precision, recall, umbral = precision_recall_curve(Y_test, ecm)

plt.plot(umbral, precision[1:], label="Precision",linewidth=5)
plt.plot(umbral, recall[1:], label="Recall",linewidth=5)
plt.title('Precision y Recall for different threshold')
plt.xlabel('Umbral')
plt.ylabel('Precision/Recall')
plt.legend()
plt.show()
```

# 5.3. Matriz de confusión

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix

umbral_fijo = .75
Y_pred = [1 if e > umbral_fijo else 0 for e in ecm]

conf_matrix = confusion_matrix(Y_test, Y_pred)

print('Confusion Matrix:')
print(conf_matrix)

precision_at_threshold = precision_score(Y_test, Y_pred)
accuracy_at_threshold = accuracy_score(Y_test, Y_pred)

print(f"\nPrecision at threshold {umbral_fijo}: {precision_at_threshold:.4f}")
print(f"Accuracy at threshold {umbral_fijo}: {accuracy_at_threshold:.4f}")

# Visualize the Confusion Matrix
plt.figure(figsize=(6, 5))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted Normal (0)', 'Predicted Fraud (1)'],
            yticklabels=['Actual Normal (0)', 'Actual Fraud (1)'])
plt.title('Confusion Matrix')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()
```

```python
# Get values from the confusion matrix:
# conf_matrix = [[TN, FP],
#                [FN, TP]]

TP_value = conf_matrix[1, 1] # True Positives
FP_value = conf_matrix[0, 1] # False Positives

print(f"True Positives (TP) = {TP_value}")
print(f"False Positives (FP) = {FP_value}")

# Manual Precision Calculation with values substituted
precision_substituted = f"{TP_value} / ({TP_value} + {FP_value})"
precision_result = TP_value / (TP_value + FP_value)

print(f"\nPrecision = {precision_substituted} = {precision_result:.4f}")
```

```python
# Get values from the confusion matrix:
# conf_matrix = [[TN, FP],
#                [FN, TP]]

TP_value = conf_matrix[1, 1] # True Positives
FN_value = conf_matrix[1, 0] # False Negatives

print(f"True Positives (TP) = {TP_value}")
print(f"False Negatives (FN) = {FN_value}")

# Manual Recall Calculation with values substituted
recall_substituted = f"{TP_value} / ({TP_value} + {FN_value})"
recall_result = TP_value / (TP_value + FN_value)

print(f"\nRecall = {recall_substituted} = {recall_result:.4f}")
```

# For fraud detection, the balance is crucial:
Minimizing False Negatives (high recall) is usually paramount: Missing real fraud is often the most expensive error. Therefore, many fraud detection systems are designed to have high recall, even if it means accepting a lower precision and dealing with more false positives.\
Managing False Positives (high precision) is also important: While less critical than missing fraud, too many false positives can overwhelm manual review teams, irritate legitimate customers, and increase operational costs.
